En effet, la NBA a changé ces dernières années : lors des phases offensives, les équipes ont davantage tendance à tirer à l'extérieur de l'arc. Par curiosité, je voulais voir concrètement l'évolution du tir à 3 points au fil des années.
Pour cette expérience, j'ai utilisé un dataset de [Kaggle](https://www.kaggle.com/datasets/mahendrakaswan/nba-3point-shooting-data-19962020) qui comprend les données des tirs à 3 points de **1996 à 2020**.

Les étapes clés de mon expérience :

### 1. Prétraitement

Avant de traiter les données, il est important d'importer les bonnes bibliothèques afin d'interagir avec nos données :


```python
import pandas as pd   # Manipuler les données du dataset
import matplotlib.pyplot as plt # Visualiser les données
```

### 2. Traitement

Les données du dataset sont au format Excel, j'ai décidé de le convertir en CSV pour simplifier la lecture des données.


```python
# Lecture des données Excel
read_set = pd.DataFrame(pd.read_excel("shootingdata.xlsx", sheet_name=1, index_col=3))

# Conversion en csv du tableau Excel
read_set.to_csv("shootingdata.csv") 
```

Rien d'exceptionnel jusque-là, mais ça devient intéressant à partir d'ici !!

Maintenant que les données sont au bon format, je les ai converties en DataFrame dans le but de les manipuler et de me familiariser avec elles en affichant les données :


```python
df = pd.DataFrame(pd.read_csv("shootingdata.csv"))
print()
print(df.head())
```

Le DataFrame n'est pas très lisible et doit être traité, j'ai effectué deux tâches :

- Renommer les colonnes

- Supprimer les colonnes vides


```python

df = pd.DataFrame(pd.read_csv("shootingdata.csv"))

# Suppression des 2 premières lignes parasites et renommage des colonnes sans titre
df = df.iloc[2:,]      
df = df.rename(columns={"Unnamed: 0.1":"3PA", "Unnamed: 0":"none", "Unnamed: 1":"years", "Unnamed: 2":"3PM","Unnamed: 4":"3P%","Unnamed: 5":"3PInTotalPoints"})
df['years'] = df['years'].str.replace(r'-.*$', '', regex=True)

# Suppression des la colonne nulle 
cleaned_df = df.drop(columns=["none"], errors="raise")

print(cleaned_df)
```

Le nettoyage a été effectué et on a bien toutes les statistiques dont on a besoin pour les visualiser!


### 3. Visualisation

On arrive à la partie la plus intéressante, **la visualisation des données**.

Les données doivent être triées afin de tracer plusieurs courbes pour chaque type de données, le dataset en comporte 4, dont j'en ai retenu 3 :

- Les tirs à 3 points réussis
- Les tentatives de tir à 3 points
- La part des tirs à 3 points dans le total des points

Elles seront affichées sur l'axe des ordonnées (y), tandis que les années seront sur l'axe des abscisses (x).


```python
threePointersMade = pd.to_numeric(cleaned_df["3PM"])
threePointersAttempts = pd.to_numeric(cleaned_df["3PA"])
threePointersInTotalPoints = pd.to_numeric(cleaned_df["3PInTotalPoints"])
years = cleaned_df["years"]
```

Après tri, on passe au tracé ! On va tracer **3 courbes** en fonction des statistiques triées précédemment, et étudier les résultats.


```python
# Tracé des trois courbes 
plt.figure(figsize=(14, 6)) 
plt.plot(years, threePointersMade, color='#2ecc71', label="3 ooints réussis ")
plt.plot(years, threePointersAttempts, color='#e74c3c', label="Tentatives de tirs à 3 points")
plt.plot(years, threePointersInTotalPoints, color='#3498db',  label="Pourcentage des 3 points au total")
plt.axvline(x="2012", color='red', linestyle='--', linewidth=1.5)
plt.text("2012", plt.ylim()[1] * 0.75, "  Début de la croissance", 
         color='grey', fontsize=9)
# Mise en forme du graphique
plt.xlabel("Années") 
plt.ylabel("%")
plt.tight_layout()
plt.title("3 points éffectuées dans la NBA")
plt.legend()
plt.grid(True)
plt.show()


```

On observe une croissance **flagrante** des tirs à 3 points, qui a doublé en une décennie. Cela peut s'expliquer par plusieurs raisons, mais la plus évidente est l'avènement de **l'ère des Golden State Warriors**, qui tiraient beaucoup plus à trois points que toutes les autres équipes de la NBA. En voyant les résultats, les différentes équipes ont **adopté** cette stratégie offensive, d'où une nette croissance à partir de 2012.
